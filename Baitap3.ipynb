```python
df_can = pd.read_excel('/content/drive/MyDrive/Môn Thầy Lạc/Canada.xlsx',
                       sheet_name='Canada by Citizenship',
                       skiprows=range(20),
)
```

```python
print(df_can.shape)
```

```python
df_can.drop(['AREA', 'REG', 'DEV', 'Type', 'Coverage'], axis=1, inplace=False)
```

```python
df_can.rename(columns={'OdName':'Country', 'AreaName':'Continent','RegName':'Region'}, inplace=True)
df_can.head()
```

```python
all(isinstance(column, str) for column in df_can.columns)
df_can.dtypes
```

```python
df_can.columns = list(map(str, df_can.columns))
all(isinstance(column, str) for column in df_can.columns)
df_can.dtypes
```

```python
df_can.set_index('Country', inplace=True)
df_can.head()
```

```python
df_can.columns = list(map(str, df_can.columns))
df_can.columns
```

```python
years_int = list(range(1980, 2014)) # list years, type = int
years_str = list(map(str, years_int)) # Convert int to str
print(years_int)
print("=" * 280)
print(years_str)
```

```python
df_can[years_str].head()
```

```python
df_can['Total'] = df_can[years_str].sum(axis=1)
df_can
```

```python
all(isinstance(column, str) for column in df_can.columns)
```

```python
drop_row = ['Total', 'Unknown']
df_can.drop(drop_row, axis=0, inplace=True, errors='ignore')
df_can
```

```python
years = list(map(str, range(1980, 2014)))
years
```

```python
df_can[years].sum(axis=1)
```

```python
df_can['Total'] = df_can[years].sum(axis=1)
df_can
```

```python
continent = df_can.groupby('Continent').sum()
continent
```

```python
continent = df_can.groupby('Continent').sum()[['Total']]
continent
```

```python
continent['Total'].plot(kind = 'pie',
                        subplots = True)
```

```python
continent = df_can.groupby('Continent').sum()
continent
```

```python
continent['2013'].plot(kind='pie')
```

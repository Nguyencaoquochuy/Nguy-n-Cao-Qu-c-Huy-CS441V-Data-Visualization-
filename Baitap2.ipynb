# Q1 Load và kiểm tra dataset

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/Môn Thầy Hà/sales_data.csv')
```

## Q2 Tạo cột mới cho tổng giá trị đơn hàng

```python
df = df.assign(total_amount = df['quantity']*df['unit_price'])
```

```python
df.assign()
```

# Q3 Lọc các đơn hàng có giá trị cao

```python
df_order_big = df[df['total_amount']>500]
```

```python
df_order_big
```

# Q4 Đếm số đơn hàng của mỗi khách hàng

```python
orders_per_customer = df.groupby('customer_id')['order_id'].count()
```

```python
orders_per_customer
```

# Q5 Tính tổng chi tiêu của mỗi khách hàng

```python
total_pay = df.groupby('customer_id').agg(total_spent=('total_amount', 'sum'))
```

```python
total_pay
```

# Q6 Phân tích doanh thu theo ngày

```python
df = pd.read_csv("/content/drive/MyDrive/Môn Thầy Hà/sales_data.csv",parse_dates=["order_date"])
df['total_amount'] = df['quantity'] * df['unit_price']
daily_revenue = (df.groupby("order_date")["total_amount"].sum().sort_values())
```

```python
daily_revenue
```

# Q7 Xếp hạng khách hàng theo mức chỉ tiêu

```python
total_pay["rank"] = total_pay["total_spent"].rank(
    ascending=False)
```
